```python
df_clean = df.copy()

none_cols = ["Alley", "BsmtQual", "BsmtCond", "BsmtExposure", "BsmtFinType1", "BsmtFinType2",
             "FireplaceQu", "GarageType", "GarageFinish", "GarageQual", "GarageCond",
             "PoolQC", "Fence", "MiscFeature", "MasVnrType"]
for c in [c for c in none_cols if c in df_clean]:
    df_clean[c] = df_clean[c].fillna("None")

# numeric 'absent' features
for c in [c for c in ["MasVnrArea", "BsmtFinSF1", "BsmtFinSF2", "BsmtUnfSF", "TotalBsmtSF",
                      "BsmtFullBath", "BsmtHalfBath", "GarageCars", "GarageArea"] if c in df_clean]:
    df_clean[c] = df_clean[c].fillna(0)

if "GarageYrBlt" in df_clean:
    df_clean["HasGarage"] = df_clean["GarageYrBlt"].notnull().astype(int)
    df_clean["GarageYrBlt"] = df_clean["GarageYrBlt"].fillna(df_clean["YearBuilt"])

if "LotFrontage" in df_clean:
    df_clean["LotFrontage"] = df_clean.groupby("Neighborhood")["LotFrontage"] \
                                      .transform(lambda s: s.fillna(s.median()))
    df_clean["LotFrontage"] = df_clean["LotFrontage"].fillna(df_clean["LotFrontage"].median())

if "Electrical" in df_clean:
    df_clean["Electrical"] = df_clean["Electrical"].fillna(df_clean["Electrical"].mode()[0])

# Any leftover gaps: mode for categoricals, median for numerics
for c in df_clean.columns[df_clean.isnull().any()]:
    if df_clean[c].dtype == "object":
        df_clean[c] = df_clean[c].fillna(df_clean[c].mode()[0])
    else:
        df_clean[c] = df_clean[c].fillna(df_clean[c].median())

print("Missing values before:", int(df.isnull().sum().sum()))
print("Missing values after :", int(df_clean.isnull().sum().sum()))
```

```python
def eta_squared(data, cat, target="SalePrice"):
    gm = data[target].mean()
    ss_between = sum(len(g) * (g[target].mean() - gm) ** 2 for _, g in data.groupby(cat))
    return ss_between / ((data[target] - gm) ** 2).sum()

rows = []
for c in cat_cols:
    groups = [g["SalePrice"].values for _, g in df_clean.groupby(c) if len(g) > 1]
    if len(groups) < 2:
        continue
    h, p = stats.kruskal(*groups)
    rows.append((c, df_clean[c].nunique(), eta_squared(df_clean, c), p))

cat_effect = pd.DataFrame(rows, columns=["feature", "n_categories", "eta_squared", "p_value"]) \
               .sort_values("eta_squared", ascending=False).reset_index(drop=True)
display(cat_effect.head(15).round(4))

plt.figure(figsize=(9, 6))
sns.barplot(x="eta_squared", y="feature", data=cat_effect.head(15), color="mediumpurple")
plt.title("Share of SalePrice variance explained by each categorical feature (η²)")
plt.show()
```

```python
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib

from sklearn.base import clone
from sklearn.model_selection import train_test_split, KFold, cross_validate, GridSearchCV, RandomizedSearchCV
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler, FunctionTransformer
from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor, VotingRegressor
from sklearn.inspection import permutation_importance
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

warnings.filterwarnings("ignore")      # hides FutureWarning / ConvergenceWarning noise
RANDOM_STATE = 42
```

```python
data = df.copy()                                   # raw housing data, NOT df_clean
assert "SalePrice" in data.columns, "df must be the training file with SalePrice"

# 1a. outliers + useless columns
data = data[~((data["GrLivArea"] > 4000) & (data["SalePrice"] < 300000))].copy()
data = data.drop(columns=["Id", "GarageYrBlt"], errors="ignore")   # GarageYrBlt duplicates YearBuilt and has a 2207 typo

# 1b. "NA" means the feature is absent
none_cols = ["Alley", "MasVnrType", "BsmtQual", "BsmtCond", "BsmtExposure", "BsmtFinType1", "BsmtFinType2",
             "FireplaceQu", "GarageType", "GarageFinish", "GarageQual", "GarageCond", "PoolQC", "Fence", "MiscFeature"]
for c in none_cols:
    data[c] = data[c].fillna("None").replace("NA", "None")

# 1c. ordered quality grades -> numbers
qual_map = {"None": 0, "Po": 1, "Fa": 2, "TA": 3, "Gd": 4, "Ex": 5}
qual_cols = ["ExterQual", "ExterCond", "BsmtQual", "BsmtCond", "HeatingQC", "KitchenQual",
             "FireplaceQu", "GarageQual", "GarageCond", "PoolQC"]
for c in qual_cols:
    data[c] = data[c].map(qual_map)

# 1d. feature engineering
data["TotalSF"]      = data["TotalBsmtSF"].fillna(0) + data["1stFlrSF"] + data["2ndFlrSF"]
data["TotalBath"]    = (data["FullBath"] + 0.5 * data["HalfBath"]
                        + data["BsmtFullBath"].fillna(0) + 0.5 * data["BsmtHalfBath"].fillna(0))
data["HouseAge"]     = (data["YrSold"] - data["YearBuilt"]).clip(lower=0)
data["RemodAge"]     = (data["YrSold"] - data["YearRemodAdd"]).clip(lower=0)
data["TotalPorchSF"] = data[["OpenPorchSF", "EnclosedPorch", "3SsnPorch", "ScreenPorch", "WoodDeckSF"]].sum(axis=1)
data["HasPool"]      = (data["PoolArea"] > 0).astype(int)
data["HasGarage"]    = (data["GarageArea"] > 0).astype(int)
data["HasFireplace"] = (data["Fireplaces"] > 0).astype(int)
data["HasBsmt"]      = (data["TotalBsmtSF"] > 0).astype(int)
data["Has2ndFlr"]    = (data["2ndFlrSF"] > 0).astype(int)

# 1e. codes that are really categories
data["MSSubClass"] = data["MSSubClass"].astype(str)
data["MoSold"]     = data["MoSold"].astype(str)

print("Shape after preparation:", data.shape)
```

```python
X = data.drop(columns="SalePrice")
y_log = np.log1p(data["SalePrice"])

X_train, X_test, y_train, y_test = train_test_split(X, y_log, test_size=0.2, random_state=RANDOM_STATE)
print("Train:", X_train.shape, " Test:", X_test.shape)
```

```python
num_features = X_train.select_dtypes(include="number").columns.tolist()
cat_features = X_train.select_dtypes(exclude="number").columns.tolist()

skew_vals  = X_train[num_features].skew()
skewed     = [c for c in num_features if abs(skew_vals[c]) > 0.75 and X_train[c].min() >= 0]
not_skewed = [c for c in num_features if c not in skewed]
print(len(num_features), "numeric |", len(cat_features), "categorical |", len(skewed), "skewed numeric")


def make_preprocessor(linear: bool):
    cat_pipe = Pipeline([
        ("imp", SimpleImputer(strategy="most_frequent")),
        ("ohe", OneHotEncoder(handle_unknown="infrequent_if_exist", min_frequency=10, sparse_output=False)),
    ])
    if linear:
        skew_pipe = Pipeline([("imp", SimpleImputer(strategy="median")),
                              ("log", FunctionTransformer(np.log1p, feature_names_out="one-to-one")),
                              ("sc", StandardScaler())])
        norm_pipe = Pipeline([("imp", SimpleImputer(strategy="median")), ("sc", StandardScaler())])
        return ColumnTransformer([("skew", skew_pipe, skewed),
                                  ("num", norm_pipe, not_skewed),
                                  ("cat", cat_pipe, cat_features)])
    return ColumnTransformer([("num", SimpleImputer(strategy="median"), num_features),
                              ("cat", cat_pipe, cat_features)])
```

```python
models = {
    "Linear Regression": Pipeline([("prep", make_preprocessor(True)),  ("model", LinearRegression())]),
    "Ridge":             Pipeline([("prep", make_preprocessor(True)),  ("model", Ridge(alpha=10))]),
    "Lasso":             Pipeline([("prep", make_preprocessor(True)),  ("model", Lasso(alpha=0.001, max_iter=20000))]),
    "ElasticNet":        Pipeline([("prep", make_preprocessor(True)),  ("model", ElasticNet(alpha=0.001, l1_ratio=0.5, max_iter=20000))]),
    "Random Forest":     Pipeline([("prep", make_preprocessor(False)), ("model", RandomForestRegressor(n_estimators=300, random_state=RANDOM_STATE))]),
    "Gradient Boosting": Pipeline([("prep", make_preprocessor(False)), ("model", GradientBoostingRegressor(
                                    n_estimators=500, learning_rate=0.05, max_depth=3, subsample=0.8, random_state=RANDOM_STATE))]),
}

# Optional extra models (only used if the libraries are installed: pip install xgboost lightgbm)
try:
    from xgboost import XGBRegressor
    models["XGBoost"] = Pipeline([("prep", make_preprocessor(False)), ("model", XGBRegressor(
        n_estimators=600, learning_rate=0.03, max_depth=3, subsample=0.8, colsample_bytree=0.6, random_state=RANDOM_STATE))])
except ImportError:
    print("xgboost not installed - skipping (optional)")
try:
    from lightgbm import LGBMRegressor
    models["LightGBM"] = Pipeline([("prep", make_preprocessor(False)), ("model", LGBMRegressor(
        n_estimators=600, learning_rate=0.03, num_leaves=8, subsample=0.8, subsample_freq=1,
        colsample_bytree=0.6, random_state=RANDOM_STATE, verbose=-1))])
except ImportError:
    print("lightgbm not installed - skipping (optional)")

cv = KFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
rows = []
for name, pipe in models.items():
    res = cross_validate(pipe, X_train, y_train, cv=cv, n_jobs=-1,
                         scoring=("neg_root_mean_squared_error", "r2"))
    rows.append({"Model": name,
                 "CV RMSE (log)": -res["test_neg_root_mean_squared_error"].mean(),
                 "CV RMSE std":   res["test_neg_root_mean_squared_error"].std(),
                 "CV R2":         res["test_r2"].mean()})
    print(f"done: {name}")

cv_table = pd.DataFrame(rows).sort_values("CV RMSE (log)").reset_index(drop=True)
display(cv_table.round(4))
```

```python
ridge_search = GridSearchCV(models["Ridge"], {"model__alpha": [1, 3, 10, 30, 100]},
                            cv=cv, scoring="neg_root_mean_squared_error", n_jobs=-1)
lasso_search = GridSearchCV(models["Lasso"], {"model__alpha": [0.0003, 0.0005, 0.001, 0.002, 0.005, 0.01]},
                            cv=cv, scoring="neg_root_mean_squared_error", n_jobs=-1)
gb_search = RandomizedSearchCV(
    models["Gradient Boosting"],
    {"model__n_estimators":     [300, 500, 800],
     "model__learning_rate":    [0.02, 0.03, 0.05, 0.08],
     "model__max_depth":        [2, 3, 4],
     "model__min_samples_leaf": [3, 5, 10],
     "model__subsample":        [0.7, 0.8, 1.0],
     "model__max_features":     [0.3, 0.5, 1.0]},
    n_iter=15, cv=cv, scoring="neg_root_mean_squared_error", n_jobs=-1, random_state=RANDOM_STATE)

for nm, s in [("Ridge", ridge_search), ("Lasso", lasso_search), ("Gradient Boosting", gb_search)]:
    s.fit(X_train, y_train)
    print(f"{nm:18s} best CV RMSE (log) = {-s.best_score_:.4f}   best params = {s.best_params_}")

ridge_best, lasso_best, gb_best = ridge_search.best_estimator_, lasso_search.best_estimator_, gb_search.best_estimator_
```

```python
tuned = {"Ridge": ridge_best, "Lasso": lasso_best, "Gradient Boosting": gb_best}
fitted = {}
for name, pipe in models.items():
    fitted[name] = tuned[name] if name in tuned else pipe.fit(X_train, y_train)

# Ensemble: average of the best linear model and the best boosting model
ensemble = VotingRegressor([("lasso", lasso_best), ("gb", gb_best)]).fit(X_train, y_train)
fitted["Ensemble (Lasso + GB)"] = ensemble


def evaluate(name, model, X_, y_log_):
    pred, actual = np.expm1(model.predict(X_)), np.expm1(y_log_)
    return {"Model": name,
            "RMSE ($)":  np.sqrt(mean_squared_error(actual, pred)),
            "MAE ($)":   mean_absolute_error(actual, pred),
            "MAPE (%)":  np.mean(np.abs(actual - pred) / actual) * 100,
            "R2":        r2_score(actual, pred),
            "RMSLE":     np.sqrt(mean_squared_error(y_log_, np.log1p(pred)))}


test_table = (pd.DataFrame([evaluate(n, m, X_test, y_test) for n, m in fitted.items()])
              .sort_values("RMSE ($)").reset_index(drop=True))
display(test_table.round(3))

best_name  = test_table.loc[0, "Model"]
best_model = fitted[best_name]
print(f"\nBest model on the hold-out set: {best_name}")
```

```python
def pdp_dollars(model, X_, feature, n_points=25):
    grid = np.unique(np.linspace(X_[feature].quantile(0.02), X_[feature].quantile(0.98), n_points))
    out = []
    for v in grid:
        tmp = X_.copy(); tmp[feature] = v
        out.append(np.expm1(model.predict(tmp)).mean())
    return grid, np.array(out)


pdp_feats = [f for f in ["OverallQual", "TotalSF", "HouseAge", "GarageCars", "TotalBath", "LotArea"] if f in X_train.columns]
fig, axes = plt.subplots(2, 3, figsize=(17, 8))
for ax, f in zip(axes.ravel(), pdp_feats):
    g, p = pdp_dollars(best_model, X_train, f)
    ax.plot(g, p, marker="o", ms=3); ax.set_title(f); ax.set_ylabel("avg predicted price ($)")
for ax in axes.ravel()[len(pdp_feats):]:
    ax.axis("off")
plt.suptitle("How the predicted price changes as each feature changes", y=1.02)
plt.tight_layout(); plt.show()
```

```python
def recommend_houses(
    data,
    max_price,
    min_area,
    min_bedrooms,
    min_bathrooms,
    min_quality=5
):
    
    recommendations = data[
        (data['SalePrice'] <= max_price) &
        (data['GrLivArea'] >= min_area) &
        (data['BedroomAbvGr'] >= min_bedrooms) &
        (data['FullBath'] >= min_bathrooms) &
        (data['OverallQual'] >= min_quality)
    ].copy()
    
    recommendations['Price_per_SqFt'] = (
        recommendations['SalePrice'] /
        recommendations['GrLivArea']
    )
    
    recommendations = recommendations.sort_values(
        by=['OverallQual', 'Price_per_SqFt'],
        ascending=[False, True]
    )
    
    return recommendations[
        [
            'SalePrice',
            'Neighborhood',
            'GrLivArea',
            'BedroomAbvGr',
            'FullBath',
            'OverallQual',
            'Price_per_SqFt'
        ]
    ]
```

```python
recommendations = recommend_houses(
    df,
    max_price=250000,
    min_area=1500,
    min_bedrooms=3,
    min_bathrooms=2,
    min_quality=6
)

recommendations.head(10)
```
